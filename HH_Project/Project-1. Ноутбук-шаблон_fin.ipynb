* Неоконченное высшее образование 2000  Балтийская государственная академия рыбопромыслового флота…
Нас будет интересовать только уровень образования.

Создайте с помощью функции-преобразования новый признак **"Образование"**, который должен иметь 4 категории: "высшее", "неоконченное высшее", "среднее специальное" и "среднее".

Выполните преобразование, ответьте на контрольные вопросы и удалите признак "Образование и ВУЗ".

Совет: обратите внимание на структуру текста в столбце **"Образование и ВУЗ"**. Гарантируется, что текущий уровень образования соискателя всегда находится в первых 2ух слов и начинается с заглавной буквы. Воспользуйтесь этим.

*Совет: проверяйте полученные категории, например, с помощью метода unique()*


```python
def get_education(text):
    if pd.isna(text):
        return None
    text = str(text)
    
    # Проверяем по шаблонам (важно: порядок важен!)
    if text.startswith('Высшее образование'):
        return 'высшее'
    elif text.startswith('Неоконченное высшее'):
        return 'неоконченное высшее'
    elif text.startswith('Среднее специальное'):
        return 'среднее специальное'
    elif text.startswith('Среднее образование'):
        return 'среднее'
    else:
        return None

df['Образование'] = df['Образование и ВУЗ'].apply(get_education)

# Удаляем признак «Образование и ВУЗ»
df.drop(columns=['Образование и ВУЗ'], inplace=True)

# Подсчёт количества соискателей по каждой категории образования
education_counts = df['Образование'].value_counts()

print("\n" + "="*54)
print("СТАТИСТИКА ПО УРОВНЮ ОБРАЗОВАНИЯ")
print("="*54)
for category, count in education_counts.items():
    percentage = (count / len(df)) * 100
    print(f"{category:25}: {count:>6} соискателей ({percentage:>5.2f}%)")

# Общая сумма
total = education_counts.sum()
print("\n" + "="*54)
print(f"ИТОГО: {total} соискателей")
```

3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
import pandas as pd
import numpy as np
import re

def experience_to_months(text):
    if pd.isna(text) or not isinstance(text, str) or text.strip() == 'Не указано':
        return np.nan
    
    # Ищем годы и месяцы с помощью регулярных выражений
    years = 0
    months = 0
    
    # Поиск годов: цифра перед словами год/года/лет
    year_match = re.search(r'(\d+)\s*(?:год|года|лет)', text)
    if year_match:
        years = int(year_match.group(1))
    
    # Поиск месяцев: цифра перед словами месяц/месяца/месяцев
    month_match = re.search(r'(\d+)\s*(?:месяц|месяца|месяцев)', text)
    if month_match:
        months = int(month_match.group(1))
    
    total = years * 12 + months
    #return total if total > 0 else np.nan
    return max(total, 0) if total > 0 else np.nan  # ← ИСПРАВЛЕНО

# Применяем
df['Опыт работы (месяц)'] = df['Опыт работы'].apply(experience_to_months)

# Удаляем исходный столбец
df.drop(columns=['Опыт работы'], inplace=True)

# Результаты
print("\n" + "="*50)
print("ОТВЕТ НА ВОПРОС")
print("="*50)
print(f"Медианный опыт работы: {df['Опыт работы (месяц)'].median():.1f} месяцев")
```

Создадим отдельные признаки **"Город"**, **"Готовность к переезду"**, **"Готовность к командировкам"**. При этом важно учесть:

* Признак **"Город"** должен содержать только 4 категории: "Москва", "Санкт-Петербург" и "город-миллионник" (их список ниже), остальные обозначьте как "другие".

    Список городов-миллионников:

    Инфорация о метро, рядом с которым проживает соискатель нас не интересует.

Выполните преобразования и удалите столбец **"Город, переезд, командировки"** из таблицы.

*Совет: обратите внимание на то, что структура текста может меняться в зависимости от указания ближайшего метро. Учите это, если будете использовать порядок слов в своей программе.*


```python
# Список городов-миллионников
million_cities = ['Новосибирск', 'Екатеринбург', 'Нижний Новгород', 'Казань', 'Челябинск', 
                  'Омск', 'Самара', 'Ростов-на-Дону', 'Уфа', 'Красноярск', 'Пермь', 
                  'Воронеж', 'Волгоград']

def extract_city(text):
    """
    Извлекает город из строки.
    Возвращает категорию: 'Москва', 'Санкт-Петербург', 'город-миллионник' или 'другие'
    """
    if pd.isna(text) or not isinstance(text, str):
        return np.nan
    
    # Берём первую часть до запятой (это город)
    city_part = text.split(',')[0].strip()
    
    # Проверяем на Москву и Санкт-Петербург
    if city_part == 'Москва':
        return 'Москва'
    elif city_part == 'Санкт-Петербург':
        return 'Санкт-Петербург'
    elif city_part in million_cities:
        return 'город-миллионник'
    else:
        return 'другие'

def extract_relocation_ready(text):
    """
    Извлекает готовность к переезду.
    Возвращает True или False.
    """
    if pd.isna(text) or not isinstance(text, str):
        return False
    
    # Ключевые слова, указывающие на готовность к переезду
    positive_keywords = ['готов к переезду', 'готова к переезду', 'хочет переехать', 
                         'хочу переехать', 'планирует переезд', 'желает переехать']
    
    # Ключевые слова, указывающие на неготовность к переезду
    negative_keywords = ['не готов к переезду', 'не готова к переезду', 
                        'не планирует переезд', 'против переезда']
    
    text_lower = text.lower()
    
    # Сначала проверяем негативные ключевые слова
    for keyword in negative_keywords:
        if keyword in text_lower:
            return False
    
    # Затем проверяем позитивные
    for keyword in positive_keywords:
        if keyword in text_lower:
            return True
    
    # Если ничего не нашли, считаем, что не готов к переезду
    return False

def extract_travel_ready(text):
    """
    Извлекает готовность к командировкам.
    Возвращает True или False.
    По умолчанию (если информация потеряна) - False.
    """
    if pd.isna(text) or not isinstance(text, str):
        return False
    
    # Ключевые слова, указывающие на готовность к командировкам
    positive_keywords = ['готов к командировкам', 'готова к командировкам', 
                        'к командировкам готов', 'к командировкам готова',
                        'готов к редким командировкам', 'готова к редким командировкам']
    
    # Ключевые слова, указывающие на неготовность к командировкам
    negative_keywords = ['не готов к командировкам', 'не готова к командировкам',
                        'к командировкам не готов', 'к командировкам не готова']
    
    text_lower = text.lower()
    
    # Проверяем негативные ключевые слова
    for keyword in negative_keywords:
        if keyword in text_lower:
            return False
    
    # Проверяем позитивные ключевые слова
    for keyword in positive_keywords:
        if keyword in text_lower:
            return True
    
    # Если информация потеряна - False
    return False

# Применяем преобразования
df['Город'] = df['Город, переезд, командировки'].apply(extract_city)
df['Готовность к переезду'] = df['Город, переезд, командировки'].apply(extract_relocation_ready)
df['Готовность к командировкам'] = df['Город, переезд, командировки'].apply(extract_travel_ready)

# Проверяем результат
print("=== РЕЗУЛЬТАТ ПРЕОБРАЗОВАНИЯ ===")
print(f"Всего записей: {len(df)}")

print("\n=== РАСПРЕДЕЛЕНИЕ ПО ГОРОДАМ ===")
city_distribution = df['Город'].value_counts()
print(city_distribution)
print(f"\nВ процентах:")
print(df['Город'].value_counts(normalize=True) * 100)

print("\n=== ГОТОВНОСТЬ К ПЕРЕЕЗДУ ===")
relocation_dist = df['Готовность к переезду'].value_counts()
print(relocation_dist)
print(f"\nГотовы к переезду: {relocation_dist.get(True, 0)} ({relocation_dist.get(True, 0)/len(df)*100:.2f}%)")
print(f"Не готовы к переезду: {relocation_dist.get(False, 0)} ({relocation_dist.get(False, 0)/len(df)*100:.2f}%)")

print("\n=== ГОТОВНОСТЬ К КОМАНДИРОВКАМ ===")
travel_dist = df['Готовность к командировкам'].value_counts()
print(travel_dist)
print(f"\nГотовы к командировкам: {travel_dist.get(True, 0)} ({travel_dist.get(True, 0)/len(df)*100:.2f}%)")
print(f"Не готовы к командировкам: {travel_dist.get(False, 0)} ({travel_dist.get(False, 0)/len(df)*100:.2f}%)")

# Результаты
print("\n" + "="*50)
print("ОТВЕТЫ НА ВОПРОСЫ")
print("="*50)

# Процент в Санкт-Петербурге (округлить до целого)
spb_pct = round((df['Город'] == 'Санкт-Петербург').mean() * 100)
print(f"Живут в Санкт-Петербурге: {spb_pct}%")

# Процент готовых и к переезду, и к командировкам
both_pct = round(((df['Готовность к переезду']) & (df['Готовность к командировкам'])).mean() * 100)
print(f"Готовы к переезду и командировкам: {both_pct}%")

# Удаляем исходный столбец
df.drop(columns=['Город, переезд, командировки'], inplace=True)
```

Теперь у нас есть вся необходимая информация для того, чтобы создать признак "ЗП (руб)" - заработная плата в рублях.

После ответа на контрольные вопросы удалите исходный столбец заработной платы "ЗП" и все промежуточные столбцы, если вы их создавали.

Итак, давайте обсудим возможный алгоритм преобразования: 
1. Перевести признак "Обновление резюме" из таблицы с резюме в формат datetime и достать из него дату. В тот же формат привести признак "date" из таблицы с валютами.
2. Выделить из столбца "ЗП" сумму желаемой заработной платы и наименование валюты, в которой она исчисляется. Наименование валюты перевести в стандарт ISO согласно с таблицей выше.
3. Присоединить к таблице с резюме таблицу с курсами по столбцам с датой и названием валюты (подумайте, какой тип объединения надо выбрать, чтобы в таблице с резюме сохранились данные о заработной плате, изначально представленной в рублях). Значение close для рубля заполнить единицей 1 (курс рубля самого к себе)
4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".


```python
# Функция для извлечения числовой части зарплаты из строки
def get_salary_num(arg):
    salary = float(arg.split(' ')[0])
    return salary

# Функция для определения валюты зарплаты
def get_salary_currency(arg):
    currency_dict = {
        'USD': 'USD', 'KZT': 'KZT',
        'грн': 'UAH', 'белруб': 'BYN',
        'EUR': 'EUR', 'KGS': 'KGS',
        'сум': 'UZS', 'AZN': 'AZN'
    }
    curr = arg.split(' ')[1].replace('.', '')
    if curr == 'руб':
        return 'RUB'
    else:
        return currency_dict[curr]

# Добавляем датасет курса валют    
rates = pd.read_csv('data/ExchangeRates.csv', sep = ',') 
# currencies_df = pd.read_csv('ExchangeRates.csv', sep=',')
rates['date'] = pd.to_datetime(rates['date'], dayfirst=True).dt.date

# Преобразование столбца 'Обновление резюме' в формат даты
df['Обновление резюме'] = pd.to_datetime(df['Обновление резюме'], dayfirst=True).dt.date

# Создание временных столбцов 'ЗП (врем)' и 'Курс (врем)' для числовой части зарплаты и валюты соответственно
df['ЗП (врем)'] = df['ЗП'].apply(get_salary_num)
df['Курс (врем)'] = df['ЗП'].apply(get_salary_currency)

# Объединение DataFrame data_resume и rates по временным столбцам 'Курс (врем)' и 'Обновление резюме'
merged = df.merge(
    rates, 
    left_on=['Курс (врем)', 'Обновление резюме'],
    right_on=['currency', 'date',], 
    how='left'
)

# Заполнение пропущенных значений столбцов 'close' и 'proportion' значениями 1
merged['close'] = merged['close'].fillna(1)
merged['proportion'] = merged['proportion'].fillna(1)

# Вычисление столбца 'ЗП (руб)' как произведения 'close' на 'ЗП (врем)' и деления на 'proportion'
df['ЗП (руб)'] = merged['close'] * merged['ЗП (врем)'] / merged['proportion']

# Удаление столбцов 'ЗП', 'ЗП (tmp)' и 'Курс (tmp)' из DataFrame df
df = df.drop(['ЗП', 'ЗП (врем)', 'Курс (врем)'], axis=1)

print("\n" + "="*50)
print("ОТВЕТ НА ВОПРОС")
print("="*50)

print(f"Желаемая медианная зарпалата: {round(df['ЗП (руб)'].median()/1000)} тыс. рублей")
```

```python
# Копируем данные
df_task = df.copy()

# Преобразуем boolean в строки (если необходимо)
if df_task['Готовность к переезду'].dtype == 'bool':
    df_task['Готовность к переезду'] = df_task['Готовность к переезду'].map({True: 'Да', False: 'Нет'})
elif df_task['Готовность к переезду'].dtype == 'object':
    # Если уже строки, но возможно с маленькой буквы
    df_task['Готовность к переезду'] = df_task['Готовность к переезду'].map(lambda x: 'Да' if str(x).lower() in ['да', 'true', '1', 'yes'] else 'Нет' if str(x).lower() in ['нет', 'false', '0', 'no'] else str(x))

if df_task['Готовность к командировкам'].dtype == 'bool':
    df_task['Готовность к командировкам'] = df_task['Готовность к командировкам'].map({True: 'Да', False: 'Нет'})
elif df_task['Готовность к командировкам'].dtype == 'object':
    df_task['Готовность к командировкам'] = df_task['Готовность к командировкам'].map(lambda x: 'Да' if str(x).lower() in ['да', 'true', '1', 'yes'] else 'Нет' if str(x).lower() in ['нет', 'false', '0', 'no'] else str(x))

# Оставляем только записи с ЗП < 1 млн руб.
df_filtered = df_task[df_task['ЗП (руб)'] < 1_000_000]

print(f"\nКоличество записей после фильтрации ЗП < 1 млн: {len(df_filtered)}")

# 2. ПРОВЕРКА НАЛИЧИЯ КАТЕГОРИИ "Да / Да"
both_yes_count = len(df_filtered[(df_filtered['Готовность к переезду'] == 'Да') & (df_filtered['Готовность к командировкам'] == 'Да')])

print(f"Количество записей с 'Да / Да': {both_yes_count}")

if both_yes_count == 0:
    print("\nВНИМАНИЕ: Нет записей с готовностью 'Да / Да'")
    print("Возможные причины:")
    print("1. В данных другие значения (например, 'TRUE/FALSE', '1/0')")
    print("2. Нет соискателей, готовых и к переезду, и к командировкам")
    print("3. Все такие записи отфильтрованы из-за ЗП < 1 млн")
    
    # Показываем все возможные комбинации
    print("\nВсе комбинации значений в данных:")
    print(df_filtered.groupby(['Готовность к переезду', 'Готовность к командировкам']).size())

# 3. ГРУППИРОВКА (С as_index=False)
grouped = df_filtered.groupby(['Готовность к переезду', 'Готовность к командировкам'], as_index=False)['ЗП (руб)'].median()

print("\nРезультат группировки:")
print(grouped)

# 4. ПОЛУЧАЕМ ОТВЕТ (С ПРОВЕРКОЙ)
# Ищем значение для "Да" / "Да"
mask = (grouped['Готовность к переезду'] == 'Да') & (grouped['Готовность к командировкам'] == 'Да')
if mask.any():
    median_both = grouped[mask]['ЗП (руб)'].values[0]
    answer = round(median_both / 1000)
else:
    # Если нет "Да/Да", ищем ближайшую категорию
    print("\nКатегория 'Да/Да' не найдена. Используем первую доступную категорию.")
    median_both = grouped['ЗП (руб)'].values[0] if len(grouped) > 0 else 0
    answer = math.ceil(median_both / 1000) if median_both > 0 else 0
    print(f"Использована категория: {grouped.iloc[0]['Готовность к переезду']} / {grouped.iloc[0]['Готовность к командировкам']}")

print(f"\nМедианная ЗП: {median_both:,.0f} руб.")
print(f"Ответ (в тысячах): {answer}")

# 5. ПОСТРОЕНИЕ ДИАГРАММЫ
if len(grouped) > 0:
    fig = px.bar(grouped, x='Готовность к переезду', y='ЗП (руб)', color='Готовность к командировкам',
        title='Медианная заработная плата в зависимости от готовности к переезду и командировкам', text='ЗП (руб)',
        barmode='group', color_discrete_sequence=px.colors.qualitative.Set2)

    fig.update_traces(texttemplate='%{text:,.0f}', textposition='outside')

    fig.update_layout(height=500, width=900, xaxis_title='Готовность к переезду', yaxis_title='Медианная заработная плата (руб.)',
        yaxis_tickformat=',.0f', legend_title='Готовность к командировкам', plot_bgcolor='white')

    fig.show()
else:
    print("Нет данных для построения диаграммы")

#fig.write_html('plotly/hh_analysis_chart_7.html')
print("График сохранён в 'plotly/hh_analysis_chart_7.html'")
```

```python
# Фильтруем данные
filtered_df = df[
    (df['ЗП (руб)'] < 1_000_000) & (df['ЗП (руб)'] > 5000) & (df['Возраст'].notna()) & (df['Образование'].notna())].copy()

print("="*70)
print("АНАЛИЗ ЗАВИСИМОСТИ ЗАРПЛАТЫ ОТ ВОЗРАСТА И ОБРАЗОВАНИЯ")
print("="*70)
print(f"\nКоличество записей для анализа: {len(filtered_df)}")

if len(filtered_df) == 0:
    print("Нет данных для анализа. Проверьте фильтры.")
    exit()

# Создаем возрастные группы
bins = [18, 25, 30, 35, 40, 45, 50, 55, 60, 100]
labels = ['18-24', '25-29', '30-34', '35-39', '40-44', '45-49', '50-54', '55-59', '60+']
filtered_df['Возрастная группа'] = pd.cut(filtered_df['Возраст'], bins=bins, labels=labels, right=False)

# Удаляем строки с NaN в возрастной группе
filtered_df = filtered_df[filtered_df['Возрастная группа'].notna()]

# 2. СВОДНАЯ ТАБЛИЦА (МЕДИАНЫ) С ОБРАБОТКОЙ ОШИБОК

# Создаем сводную таблицу
pivot_table = filtered_df.pivot_table(values='ЗП (руб)', index='Образование', columns='Возрастная группа', aggfunc='median', fill_value=0)

# Если таблица пустая, выходим
if pivot_table.empty:
    print("Сводная таблица пуста. Нет данных для анализа.")
    exit()

# Сортируем образование (только те, что есть в данных)
education_order = ['Среднее', 'Среднее специальное', 'Неполное высшее', 'Высшее', 'Магистр', 'Кандидат наук', 'Доктор наук']
existing_edu = [edu for edu in education_order if edu in pivot_table.index]
if len(existing_edu) == 0:
    existing_edu = list(pivot_table.index)
pivot_table = pivot_table.loc[existing_edu]

# Сортируем возрастные группы (только те, что есть)
existing_age = [age for age in labels if age in pivot_table.columns]
if len(existing_age) == 0:
    existing_age = list(pivot_table.columns)
pivot_table = pivot_table[existing_age]

print("\n" + "="*70)
print("СВОДНАЯ ТАБЛИЦА: Медианная зарплата (руб)")
print("="*70)
print(pivot_table.round(0).to_string())

# 3. ТЕПЛОВАЯ КАРТА

if not pivot_table.empty and pivot_table.shape[0] > 0 and pivot_table.shape[1] > 0:
    # Убираем строки и столбцы, где все значения = 0
    pivot_table_nonzero = pivot_table.loc[(pivot_table != 0).any(axis=1), (pivot_table != 0).any(axis=0)]
    
    if not pivot_table_nonzero.empty:
            fig_heatmap = px.imshow(pivot_table_nonzero.values, x=pivot_table_nonzero.columns, y=pivot_table_nonzero.index,
            text_auto='.0f', aspect='auto', color_continuous_scale='Viridis',
            title='Тепловая карта: Медианная зарплата в зависимости от возраста и образования',
            labels={'x': 'Возрастная группа', 'y': 'Уровень образования', 'color': 'Зарплата (руб)'})
            fig_heatmap.update_layout(height=600, width=1000, xaxis_title="Возрастная группа", yaxis_title="Уровень образования")
        
            fig_heatmap.show()
    else:
        print("\nНет данных для построения тепловой карты (все значения нулевые)")
else:
    print("\nНет данных для построения тепловой карты")


#fig_heatmap.write_html('plotly/hh_analysis_chart_8.html')
print("График сохранён в 'plotly/hh_analysis_chart_8.html'")

# 4. АНАЛИЗ КАРЬЕРНОГО РОСТА

growth_analysis = {}

for edu in pivot_table.index:
    salaries = pivot_table.loc[edu]
    # Убираем нулевые значения
    salaries_nonzero = salaries[salaries > 0]
    
    if len(salaries_nonzero) >= 2:
        max_salary = salaries_nonzero.max()
        min_salary = salaries_nonzero.min()
        
        # Находим возрастные группы
        max_age = salaries_nonzero.idxmax()
        min_age = salaries_nonzero.idxmin()
        
        # Находим индексы для расчета интенсивности
        age_list = list(salaries_nonzero.index)
        if len(age_list) >= 2:
            start_idx = age_list.index(min_age)
            end_idx = age_list.index(max_age)
            steps = abs(end_idx - start_idx)
            intensity_per_step = (max_salary - min_salary) / steps if steps > 0 else 0
        else:
            intensity_per_step = 0
        
        # Относительный рост
        relative_growth = max_salary / min_salary if min_salary > 0 else 0
        
        growth_analysis[edu] = {
            'Начальная зарплата': min_salary,
            'Пиковая зарплата': max_salary,
            'Возраст начала': min_age,
            'Возраст пика': max_age,
            'Абсолютный рост': max_salary - min_salary,
            'Рост (в разах)': relative_growth,
            'Интенсивность (руб/шаг)': intensity_per_step
        }
        
# 5. ОПРЕДЕЛЕНИЕ САМОГО БЫСТРОГО РОСТА

if len(growth_analysis) > 0:
    # По интенсивности (руб/шаг) - это основной критерий
    best_intensity = max(growth_analysis.items(), key=lambda x: x[1]['Интенсивность (руб/шаг)'])        
else:
    print("\nНедостаточно данных для определения карьерного роста")
    
    # Показываем, что есть в данных
    print("\nДоступные данные в сводной таблице:")
    print(pivot_table.round(0).to_string())

# 6. ГРАФИК КАРЬЕРНОГО РОСТА

if len(pivot_table.index) > 0 and len(pivot_table.columns) > 0:
    fig_line = go.Figure()
    
    for edu in pivot_table.index:
        salaries = pivot_table.loc[edu]
        salaries_nonzero = salaries[salaries > 0]
        
        if len(salaries_nonzero) > 1:
            fig_line.add_trace(go.Scatter(x=salaries_nonzero.index, y=salaries_nonzero.values, mode='lines+markers', name=edu,
                line=dict(width=2), marker=dict(size=8)))
    
    if len(fig_line.data) > 0:
        fig_line.update_layout(title='Карьерный рост: изменение зарплаты с возрастом', xaxis_title='Возрастная группа',
            yaxis_title='Медианная заработная плата (руб)',height=600, width=1000, legend_title='Уровень образования')
        fig_line.show()
    else:
        print("\nНедостаточно данных для построения графика карьерного роста")

#fig_line.write_html('plotly/hh_analysis_chart_9.html')
print("График сохранён в 'plotly/hh_analysis_chart_9.html'")
```

```python
# Создаём папку для сохранения графиков

# Функция распределения
def age_z_score(df, feature, log_scale=False):
    if log_scale:
        x = np.log(df[feature])
    else:
        x = df[feature]
    mu = x.mean()
    sigma = x.std()
    lower_bound = mu - 3 * sigma
    upper_bound = mu + 4 * sigma
    outliers = df[(x < lower_bound) | (x > upper_bound)]
    cleaned = df[(x >= lower_bound) & (x <= upper_bound)]
    return outliers, cleaned

# Применяем функцию распределения к возрасту
outliers, cleaned = age_z_score(df, 'Возраст', log_scale=True)
#print(f'Число выбросов по методу z-отклонения: {outliers.shape[0]}')

# Расчёт логарифма возраста
log_age = np.log(df['Возраст'])

# Создаём plotly гистограмму
fig = px.histogram(
    x=log_age,
    nbins=20,
    title='Логарифмическое распределение возраста соискателей',
    labels={'x': 'log(Возраст)', 'y': 'Количество'},
    color_discrete_sequence=['green']
)

# Добавляем вертикальные линии (границы)
mu = log_age.mean()
sigma = log_age.std()

fig.add_vline(x=mu, line_width=2, line_color='black', 
              annotation_text=f'μ = {mu:.2f}', annotation_position='top')
fig.add_vline(x=mu + 4 * sigma, line_width=2, line_dash='dash', line_color='black',
              annotation_text=f'μ+4σ = {mu + 4*sigma:.2f}', annotation_position='top')
fig.add_vline(x=mu - 3 * sigma, line_width=2, line_dash='dash', line_color='black',
              annotation_text=f'μ-3σ = {mu - 3*sigma:.2f}', annotation_position='bottom')

fig.show()

#fig.write_html('plotly/hh_analysis_chart_14.html')
print("График сохранён в 'plotly/hh_analysis_chart_14.html'")
print()
print(f'Число выбросов по методу z-отклонения: {outliers.shape[0]}')
# Расчёт асимметрии 
print("Асимметрия: " + str(round(log_age.skew(),3)))

# таблица с выбросами
display(outliers)

# Удаляем строки
df = df.drop(index=outliers.index)
```
