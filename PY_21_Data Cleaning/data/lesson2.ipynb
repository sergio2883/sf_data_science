```python
# Загрузка и очистка данных
data = pd.read_csv('diabetes_data.csv')
data_cleaned = data.drop_duplicates()

print(f"Исходное количество записей: {len(data)}")
print(f"Количество дубликатов: {len(data) - len(data_cleaned)}")
print(f"Записей после удаления дубликатов: {len(data_cleaned)}")
```

```python
n = len(data)

non_info = []
for col in data.columns:
    most_frequent_ratio = (data[col] == data[col].mode()[0]).sum() / n
    unique_ratio = data[col].nunique() / n
    
    if most_frequent_ratio >= 0.95 or unique_ratio >= 0.95:
        non_info.append(col)

print("Неинформативные признаки:")
for feature in non_info:
    print(feature)
```

```python
# Шаг 1: Удаление дубликатов
print("="*60)
print("ШАГ 1: УДАЛЕНИЕ ДУБЛИКАТОВ")
print("="*60)
print(f"Исходное количество записей: {len(data)}")
data_no_duplicates = data.drop_duplicates()
print(f"После удаления дубликатов: {len(data_no_duplicates)}")
print(f"Удалено дубликатов: {len(data) - len(data_no_duplicates)}")

# Шаг 2: Удаление неинформативных признаков
print("\n" + "="*60)
print("ШАГ 2: УДАЛЕНИЕ НЕИНФОРМАТИВНЫХ ПРИЗНАКОВ")
print("="*60)

n = len(data_no_duplicates)
non_informative = []

for col in data_no_duplicates.columns:
    most_frequent_ratio = (data_no_duplicates[col] == data_no_duplicates[col].mode()[0]).sum() / n
    unique_ratio = data_no_duplicates[col].nunique() / n
    
    print(f"\n{col}:")
    print(f"  Наиболее частое значение встречается в {most_frequent_ratio*100:.2f}% случаев")
    print(f"  Уникальных значений: {data_no_duplicates[col].nunique()} ({unique_ratio*100:.2f}%)")
    
    if most_frequent_ratio >= 0.95 or unique_ratio >= 0.95:
        non_informative.append(col)
        print(f"  >>> ПРИЗНАК НЕИНФОРМАТИВНЫЙ - будет удален")

print(f"\nНеинформативные признаки: {non_informative}")
data_cleaned = data_no_duplicates.drop(columns=non_informative)

print(f"\nПризнаки после удаления: {list(data_cleaned.columns)}")
print(f"Размер очищенных данных: {data_cleaned.shape}")

# Шаг 3: Сохранение очищенных данных для дальнейшей работы
data_cleaned.to_csv('diabetes_data_cleaned.csv', index=False)
print("\n" + "="*60)
print("ОЧИЩЕННЫЕ ДАННЫЕ СОХРАНЕНЫ В 'diabetes_data_cleaned.csv'")
print("="*60)

# Вывод первых строк очищенных данных
print("\nПервые 5 строк очищенных данных:")
print(data_cleaned.head())

# Базовая информация об очищенных данных
print("\n" + "="*60)
print("ИНФОРМАЦИЯ ОБ ОЧИЩЕННЫХ ДАННЫХ")
print("="*60)
print(data_cleaned.info())
```

```python
# Удаление столбцов с >30% пропусков
high_missing = [col for col in data.columns if data[col].isnull().sum()/len(data) > 0.3]
data = data.drop(columns=high_missing)

print(f"Количество признаков: {len(data.columns)}")
print(f"Признаки: {list(data.columns)}")
```

```python
# Удаление строк с >2 пропусками
data_cleaned = data[data.isnull().sum(axis=1) <= 2]

print(f"Результат: {len(data_cleaned)} записей")
```

```python
# 6. Замена пропусков на медиану
for col in data.columns:
    if data[col].isnull().any():
        data[col] = data[col].fillna(data[col].median())

# 7. Среднее значение SkinThickness
result = round(data['SkinThickness'].mean(), 1)
print(f"Среднее значение в столбце SkinThickness: {result}")
```

```python
# Поиск выбросов в SkinThickness методом IQR
x = data['SkinThickness']
Q1 = x.quantile(0.25)
Q3 = x.quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers_count = ((x < lower) | (x > upper)).sum()

print(f"Количество выбросов в SkinThickness: {outliers_count}")
```

```python
# Поиск выбросов в SkinThickness методом z-отклонения (3 сигмы)
x = data['SkinThickness']
mean = x.mean()
std = x.std()
lower = mean - 3 * std
upper = mean + 3 * std
outliers_count = ((x < lower) | (x > upper)).sum()

print(f"Количество выбросов в SkinThickness (метод z-отклонения): {outliers_count}")
```

```python
# Анализ выбросов в DiabetesPedigreeFunction
x = data['DiabetesPedigreeFunction']

# IQR на исходных данных
Q1 = x.quantile(0.25)
Q3 = x.quantile(0.75)
IQR = Q3 - Q1
outliers_original = ((x < (Q1 - 1.5*IQR)) | (x > (Q3 + 1.5*IQR))).sum()

# IQR на логарифмированных данных
x_log = np.log(x)
Q1_log = x_log.quantile(0.25)
Q3_log = x_log.quantile(0.75)
IQR_log = Q3_log - Q1_log
outliers_log = ((x_log < (Q1_log - 1.5*IQR_log)) | (x_log > (Q3_log + 1.5*IQR_log))).sum()

difference = outliers_original - outliers_log
print(f"Выбросов на исходных данных: {outliers_original}")
print(f"Выбросов на логарифмированных данных: {outliers_log}")
print(f"Разница: {difference}")
```
