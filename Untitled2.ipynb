```python
import pandas as pd
import plotly
import plotly.express as px
import matplotlib.pyplot as plt

plt.style.use('default')

covid_data = pd.read_csv(r'D:\IDE\github\sf_data_science\PY_20_Visualization\data\covid_data.csv', sep=',')
vaccinations_data = pd.read_csv(r'D:\IDE\github\sf_data_science\PY_20_Visualization\data\country_vaccinations.csv', sep=',')

covid_data = covid_data.groupby(
    ['date', 'country'], 
    as_index=False
)[['confirmed', 'deaths', 'recovered']].sum()

covid_data['date'] = pd.to_datetime(covid_data['date'])
covid_data['active'] = covid_data['confirmed'] - covid_data['deaths'] - covid_data['recovered']
covid_data = covid_data.sort_values(by=['country', 'date'])
covid_data['daily_confirmed'] = covid_data.groupby('country')['confirmed'].diff()
covid_data['daily_deaths'] = covid_data.groupby('country')['deaths'].diff()
covid_data['daily_recovered'] = covid_data.groupby('country')['recovered'].diff()

vaccinations_data = vaccinations_data[
    ['country', 'date', 'total_vaccinations', 
     'people_vaccinated', 'people_vaccinated_per_hundred',
     'people_fully_vaccinated', 'people_fully_vaccinated_per_hundred',
     'daily_vaccinations', 'vaccines']
]

vaccinations_data['date'] = pd.to_datetime(vaccinations_data['date'])

covid_df = pd.merge(covid_data, vaccinations_data, 
                    left_on=['date', 'country'], 
                    right_on=['date', 'country'], 
                    how='left')            


# Создаём признаки death_rate и recover_rate
covid_df['death_rate'] = (covid_df['deaths'] / covid_df['confirmed']) * 100
covid_df['recover_rate'] = (covid_df['recovered'] / covid_df['confirmed']) * 100
```

```python
line_data = covid_df.groupby('date', as_index=False).sum()
fig = px.line(
    data_frame=line_data, #DataFrame
    x='date', #ось абсцисс
    y=['confirmed', 'recovered', 'deaths', 'active'], #ось ординат
    height=500, #высота
    width=1000, #ширина
    title='Confirmed, Recovered, Deaths, Active cases over Time' #заголовок
)
fig.show()
```

```python
bar_data = covid_df.groupby(
    by='country',
    as_index=False
)[['recover_rate']].mean().round(2).nlargest(10, columns=['recover_rate'])

#строим график
fig = px.bar(
    data_frame=bar_data, #датафрейм
    x="country", #ось x
    y="recover_rate", #ось y
    color='country', #расцветка в зависимости от страны
    text = 'recover_rate', #текст на столбцах
    orientation='v', #ориентация графика
    height=500, #высота
    width=1000, #ширина
    title='Top 10 Countries for Recovery Rate' #заголовок
)

#отображаем его
fig.show()
```

```python
#считаем среднее ежедневно фиксируемое количество выздоровевших по странам
treemap_data = covid_df.groupby(
    by='country',
    as_index=False
)[['daily_recovered']].mean()

#строим график
fig = px.treemap(
    data_frame=treemap_data, #DataFrame
    path=['country'], #категориальный признак, для которого строится график
    values='daily_recovered', #параметр, который сравнивается
    height=500, #высота
    width=1000, #ширина
    title='Daily Recovered Cases by Country' #заголовок
)

#отображаем график
fig.show()
```

```python
#преобразуем даты в строковый формат
choropleth_data = covid_df.sort_values(by='date')
choropleth_data['date'] = choropleth_data['date'].astype('string')

#строим график
fig = px.choropleth(
    data_frame=choropleth_data, #DataFrame
    locations="country", #столбец с локациями
    locationmode = "country names", #режим сопоставления локаций с базой Plotly
    color="confirmed", #от чего зависит цвет
    animation_frame="date", #анимационный бегунок
    range_color=[0, 30e6], #диапазон цвета
    title='Global Spread of COVID-19', #заголовок
    width=800, #ширина
    height=500, #высота
    color_continuous_scale='Reds' #палитра цветов
)

#отображаем график
fig.show()
```

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Агрегируем ежедневные вакцинации по всему миру
world_vaccinations = covid_df.groupby('date')['daily_vaccinations'].sum().reset_index()

# Преобразуем дату в формат datetime
world_vaccinations['date'] = pd.to_datetime(world_vaccinations['date'])

# Сортируем по дате
world_vaccinations = world_vaccinations.sort_values('date')

# Переводим в миллионы для удобства
world_vaccinations['daily_vaccinations_millions'] = world_vaccinations['daily_vaccinations'] / 1_000_000
```

```python
# Вариант 1: С помощью plotly.express
fig = px.line(world_vaccinations, 
              x='date', 
              y='daily_vaccinations_millions',
              title='Ежедневное количество вакцинированных в мире',
              labels={'date': 'Дата', 'daily_vaccinations_millions': 'Daily vaccinations (млн человек)'},
              line_shape='linear')

# Настройка внешнего вида
fig.update_traces(line=dict(color='green', width=2.5))
fig.update_layout(
    xaxis_title='Дата',
    yaxis_title='Количество вакцинированных (млн)',
    hovermode='x unified',
    template='plotly_white'
)

# Добавляем вертикальную линию и точку на 28 февраля 2021
target_date = '2021-02-28'
target_date_dt = pd.to_datetime(target_date)

# Проверяем, есть ли данные за эту дату
if target_date_dt in world_vaccinations['date'].values:
    target_value = world_vaccinations[world_vaccinations['date'] == target_date_dt]['daily_vaccinations_millions'].values[0]
    
    # Добавляем точку
    fig.add_trace(go.Scatter(
        x=[target_date_dt],
        y=[target_value],
        mode='markers+text',
        marker=dict(size=12, color='red', symbol='circle'),
        text=[f'{target_value:.1f} млн'],
        textposition='top center',
        name=f'28 февраля 2021',
        showlegend=True
    ))
    
    # Добавляем вертикальную линию
    fig.add_vline(x=target_date_dt, line_dash="dash", line_color="red", opacity=0.5)

fig.show()
```

```python
# Шаг 4: Извлекаем значение для Японии на 24 марта 2021
target_country = 'Japan'
target_date = '2021-03-24'

# Фильтруем данные
japan_target = covid_df[
    (covid_df['country'] == target_country) & 
    (covid_df['date'] == target_date)
]

if not japan_target.empty:
    total_vac = japan_target['total_vaccinations'].values[0]
    
    # Проверяем, не NaN ли значение
    if pd.notna(total_vac):
        # Переводим в тысячи
        total_vac_thousands = total_vac / 1000
        
        # Округляем до целого
        answer = round(total_vac_thousands)
        
        print(f"\n{'='*50}")
        print(f"📍 Страна: {target_country}")
        print(f"📅 Дата: {target_date}")
        print(f"💉 Всего вакцинаций: {total_vac:,.0f}")
        print(f"📊 В тысячах: {total_vac_thousands:.2f}")
        print(f"\n✅ ОТВЕТ: {answer} тысяч")
        print(f"{'='*50}")
    else:
        print(f"Значение total_vaccinations для {target_country} на {target_date} отсутствует (NaN)")
else:
    print(f"Данные для {target_country} на {target_date} не найдены")
    
    # Проверяем, какие даты есть для Японии
    japan_all = covid_df[covid_df['country'] == 'Japan']
    if not japan_all.empty:
        print(f"\nДоступные даты для Японии:")
        print(f"  - С: {japan_all['date'].min()}")
        print(f"  - По: {japan_all['date'].max()}")
        
        # Находим ближайшую дату
        target_dt = pd.to_datetime(target_date)
        japan_all['date_diff'] = abs(japan_all['date'] - target_dt)
        closest = japan_all.loc[japan_all['date_diff'].idxmin()]
        print(f"\nБлижайшая доступная дата: {closest['date'].strftime('%Y-%m-%d')}")
        print(f"Значение на эту дату: {closest['total_vaccinations']:,.0f} вакцинаций")
```
